```python
import numpy as np
```

```python
N_arr = np.hstack([np.arange(1, 10),np.arange(10, 110, 10)])
```

```python
ratios = []
for n in N_arr:
    dat = np.loadtxt(f'raw_data/fsc_sx=1.0_N={n}.txt', dtype=complex)
    ratios.append (dat[-1,-1] / n)

np.savetxt('fsc_sx=1.0.txt', np.column_stack((N_arr, ratios)))
```
